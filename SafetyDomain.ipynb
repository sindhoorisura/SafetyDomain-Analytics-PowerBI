# Customer_name , Delivery_Days , Service_type, Customer_rating  Columns have non-null Values.
# Customer_name will be mapped as per Customer_ID else "Customer Unknown", Service_Type-Unknown, Customer_Rating and Delivery_Days will be replaced with zero if Nulls


```python
df["Customer_Name"] = (
    df.groupby("Customer_ID")["Customer_Name"]
      .transform(lambda x: x.ffill().bfill())
)
df["Customer_Name"].isnull().sum()
df["Customer_Name"]
df.isnull().sum()
```

```python
df["Service_Type"] = df["Service_Type"].str.replace("None", "Unknown")
df["Customer_Rating"] = df["Customer_Rating"].fillna(0)
df["Delivery_Days"] = df["Delivery_Days"].fillna(0)
df.drop_duplicates()
df.info()
```

```python
df.isnull().sum()
```

```python
df["Service_Type"] = df["Service_Type"].fillna( "Unknown")
df.isnull().sum()
```

```python
df.groupby("Region")["Revenue"].sum()
df.groupby("Region")["Revenue"].mean()
```
